```python
from keras.datasets import imdb
```

```python
from keras import Sequential
```

```python
from keras.layers import Dense ,SimpleRNN ,Embedding ,Flatten
```

```python
(X_train,y_train),(X_test,y_test)=imdb.load_data()
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
X_train = pad_sequences(X_train, padding='post', maxlen=50)
X_test = pad_sequences(X_test, padding='post', maxlen=50)
```

```python
model = Sequential()

model.add(SimpleRNN(32, input_shape=(50, 1), return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_data=(X_test, y_test)
)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred = (y_pred >= 0.5).astype(int)
```

```python
print("Predicted:", y_pred[:10].flatten())
print("Actual:   ", y_test[:10])
```

```python
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences

def predict_sentiment(text):
    # Convert text into numerical sequence
    sequence = tokenizer.texts_to_sequences([text])

    # Make sequence length 50
    padded = pad_sequences(sequence, maxlen=50, padding='post')

    # Predict sentiment
    prediction = model.predict(padded, verbose=0)[0][0]

    if prediction >= 0.5:
        print(f"Positive 😊 (Score: {prediction:.2f})")
    else:
        print(f"Negative 😞 (Score: {prediction:.2f})")
```

```python
predict_sentiment("Dhoni is the greatest")
predict_sentiment("This movie is terrible")
predict_sentiment("Modi ji ki jai")
predict_sentiment("I hate this")
```
